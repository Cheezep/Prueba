```python
from datetime import datetime
import ipywidgets as widgets
from IPython.display import Image, clear_output, display

# Definir fecha y hora límite: 11 de Septiembre de 2026 a las 5:30 PM (17:30)
fecha_limite = datetime(2026, 9, 11, 17, 30)

slides = [
    ("Si sí es, continúa leyendo...", "teto2.jpg"),
    ("Hola, gracias por ser mi skibidi best friend", "seal.jpg"),
    (
        (
            "Aprecio mucho nuestra amistad, pero ya no puedo ocultar lo que"
            " siento... (Estoy nervioso 💀)"
        ),
        "cuerpo.jpg",
    ),
    (
        (
            "Eres una persona increíble: inteligente, divertida y un poco"
            " rara...\n(me da mucho miedo el sticker de Sonic mostrando sus"
            " patas)"
        ),
        "skel.jpg",
    ),
    (
        (
            "Aún así te quiero tal y como eres.\n(Tampoco es que yo sea la"
            " persona más normal del mundo XD)"
        ),
        "perritos.jpg",
    ),
    (
        "Y lamentablemente, no tuve el valor de decirte esto en persona...",
        "gatitos.jpg",
    ),
    (
        (
            "Pero me armé de valor para decirte esto en este formato. (La"
            " intención es lo que cuenta)"
        ),
        "gato.jpg",
    ),
    (
        (
            "Y antes de que vayas corriendo a la policía para ponerme una"
            " orden de alejamiento...\nTranquila, yo haría lo mismo en tu"
            " posición"
        ),
        "policia.jpg",
    ),
    (
        (
            "Pero acuérdate que ahora estoy en UK, así que tranquila, no te"
            " puedo matar"
        ),
        "cato.jpg",
    ),
    (
        (
            "Bueno, regresando al tema y siendo serios...\nMe armo de valor"
            " para preguntarte si...\n¿Querrías ser la hipotenusa de mi"
            " cateto? (perdón, no se me ocurrió nada mejor)"
        ),
        "benzema.jpg",
    ),
]

actual = 0
out = widgets.Output()

def iniciar_presentacion():
    mostrar_slide(0)

def mostrar_slide(index):
    with out:
        clear_output(wait=True)
        texto, img = slides[index]
        print(texto + "\n")
        display(Image(filename=img, width=300))
        print("\n")

        if index < len(slides) - 1:
            btn = widgets.Button(
                description="Siguiente ➔",
                button_style="info",
                layout=widgets.Layout(width="auto"),
            )
            btn.on_click(al_hacer_clic)
            display(btn)
        else:
            btn_si = widgets.Button(
                description="¡Sí!",
                button_style="success",
                layout=widgets.Layout(width="auto"),
            )
            btn_no = widgets.Button(
                description="No",
                button_style="danger",
                layout=widgets.Layout(width="auto"),
            )

            btn_si.on_click(respuesta_si)
            btn_no.on_click(respuesta_no)

            box = widgets.HBox([btn_si, btn_no])
            display(box)

def al_hacer_clic(b):
    global actual
    actual += 1
    mostrar_slide(actual)

def respuesta_si(b):
    with out:
        clear_output(wait=True)
        print("¡YAY!\n")
        print(
            "Mándame por WhatsApp 10 emojis de la bandera de Israel y un emoji"
            " del dedo medio 🇮🇱🖕"
        )
        print("para saber que ahora Israel me tiene vigilado.\n")
        print(
            "No sé qué tan sano sea una relación a distancia, pero te prometo"
            " que haré lo mejor que pueda para hacerte feliz.\n"
        )
        display(Image(filename="dog.jpg", width=350))

def respuesta_no(b):
    with out:
        clear_output(wait=True)
        print("Tranquila,\n")
        print(
            "Sé que me ves como solo un amigo o feto, y está bien. (Aunque creo"
            " que ya arruiné nuestra amistad con esto XD)\n"
        )
        print(
            "Aún podemos hablar como siempre o puedes bloquearme, sea como sea"
            " no tienes por qué justificarte, lo entiendo.\n"
        )
        display(Image(filename="waza.jpg", width=300))

def denegar_acceso():
    with out:
        clear_output(wait=True)
        print("ACCESO DENEGADO \n")
        print(
            "Dijiste que aún no es la hora o el reloj del sistema detectó que"
            " todavía no es 11 de septiembre a las 5:30 PM..."
        )
        print("Regresa cuando sea la fecha y hora indicadas.")

# --- PASO 2: PREGUNTA Y VERIFICACIÓN ---
def verificar_fecha_inicial():
    with out:
        clear_output(wait=True)
        print("PREGUNTA DE SEGURIDAD \n")
        print(
            "¿Estás leyendo esto DESPUÉS del 11 de septiembre a las 5:30 PM"
            " (17:30)?\n"
        )

        btn_si_despues = widgets.Button(
            description="Sí, ya es después de la fecha/hora",
            button_style="primary",
            layout=widgets.Layout(width="auto"),
        )
        btn_no_antes = widgets.Button(
            description="No, todavía no es la hora",
            button_style="warning",
            layout=widgets.Layout(width="auto"),
        )

        def al_responder_si(b):
            if datetime.now() < fecha_limite:
                denegar_acceso()
            else:
                iniciar_presentacion()

        def al_responder_no(b):
            denegar_acceso()

        btn_si_despues.on_click(al_responder_si)
        btn_no_antes.on_click(al_responder_no)

        display(widgets.HBox([btn_si_despues, btn_no_antes]))

# --- PASO 1: ADVERTENCIA INICIAL CON TETO MOLESTO ---
def advertencia_inicial():
    with out:
        clear_output(wait=True)
        print(
            "Antes de seguir leyendo esto:\n¿Es 11 de septiembre o alguna fecha"
            " posterior?\nSi no lo es, no leas esto (te vas a arrepentir).\n"
        )
        display(Image(filename="teto.jpg", width=300))
        print("\n")

        btn_entendido = widgets.Button(
            description="Siguiente ➔",
            button_style="info",
            layout=widgets.Layout(width="auto"),
        )
        btn_entendido.on_click(lambda b: verificar_fecha_inicial())
        display(btn_entendido)

# Iniciar flujo
display(out)
advertencia_inicial()
```
